```python
def zip_(list1, list2):
    zipped = []
    if len(list1) >= len(list2):
        len_ = len(list2)
    else:
        len_ = len(list1)
    for i in range(len_):
        zipped.append((list1[i], list2[i]))
    return zipped

a, b = [], [1, 3, 5, 7]
zip_(a, b)
```

```python
def final_balance(init_sum,interest_rate,years,round_num=2):
    for i in range(years):
        init_sum += init_sum * interest_rate/100
    return round(init_sum, round_num)

final_balance(600, 25, 5,-2)
```

```python
def print_array(answer):
    print("###")
    print(answer )
    print("###")
def math_task(data):
    answer = []
    # возводим в третью степень
    for elem in data:
        answer += [elem ** 3]
    print_array(answer)
    # берем остаток от деления на 7
    for i in range(len(answer)):
        answer[i] = answer[i] % 7
    print_array(answer)
    # прибавляем к остатку изначальный массив
    for i in range(len(answer)):
        answer[i] = answer[i] + data[i]
    print_array(answer)
    # возвращаем результат
    return answer

math_task([1,2,3,4])
```

```python
def sum_as_ints(array):
    total = 0
    for elem in array:
        try:
            elem = int(elem)
        except ValueError:
            continue
        total += elem
    return total

sum_as_ints(['3.14', '2.15', '2,2', "q", "a"])
```

```python
def reversed_(array):
    return array[::-1]

if reversed_(reversed_([1, 2, 3])) == [1, 2, 3]:
    print("Все хорошо")
else:
    raise RuntimeError("Ошибка, после обращения дважды не получается исходный массив!")
```

```python
def find_substr(substr,str_):
    start=str_.find(substr)
    return (start, start + len(substr))

find_substr("ма", "маленькая машина")
```

```python
def check_string(string):
    result = True
    if string != string.strip():
        result = False
    elif string[0:1].upper()+string[1:].lower() != string:
        result = False
    elif string[-1] != '.':
        result = False
    return result

check_string(' В этом году будет особенно теплое море')
```
