```python
playstore.columns = playstore.columns.str.lower().str.replace(' ', '_')
playstore
```

```python
unique_playstore = playstore.drop_duplicates(subset='app').reset_index(drop=True)
# unique_playstore
type_counts = unique_playstore['type'].value_counts()

free_share = type_counts['Free'] / type_counts.sum()
round(free_share, 2)
```

```python
education_playstore = unique_playstore[(unique_playstore['category'] == 'EDUCATION') & (unique_playstore['reviews'] > 1000)].reset_index(drop=True)
education_playstore
```

```python
playstore['price']=playstore['price'].str.replace('$', '').astype(float)

unique_playstore['price'] = (
    unique_playstore['price']
    .str.replace('$', '', regex=False)
    .astype(float)
)
```

```python

result = (pd.pivot_table(
    unique_playstore,
    index=['category', 'type'],
    values=['price', 'rating', 'reviews'],
    aggfunc='mean'
    )
    .rename(columns={'price': 'mean_price', 'rating': 'mean_rating', 'reviews': 'mean_reviews'})
    .round({'mean_price': 2, 'mean_rating': 1, 'mean_reviews': 2})
)

result.to_csv('result_last2.csv', sep=',')
result
```

```python
unique_playstore['price'] = pd.to_numeric(
    unique_playstore['price'].astype(str).str.replace('$', '', regex=False),
    errors='coerce'
)

result = (
    unique_playstore.pivot_table(
        index=['category', 'type'],
        values=['price', 'rating', 'reviews'],
        aggfunc='mean'
    )
    .rename(columns={
        'price': 'mean_price',
        'rating': 'mean_rating',
        'reviews': 'mean_reviews'
    })
    .reset_index()
    [['mean_price', 'mean_rating', 'mean_reviews', 'category', 'type']]
    .round({'mean_price': 2, 'mean_rating': 2, 'mean_reviews': 2})
)

result.to_csv('result_last1.csv', sep=',', index=False)
result
```
